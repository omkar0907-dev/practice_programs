```python
# Program 1 : Write a program to print the 4th element from first and 4th element from last in a tuple
nums = (10, 20, 30, 40, 50, 60, 70, 80, 90)
print(f"Tuple : {nums}")
print(f"4th element from first : {nums[3]}")
print(f"4th element from last : {nums[-4]}")
```

```python
# Program 2 : Write a program to check whether an element exists in a tuple or not
nums = (10, 20, 30, 40, 50)
target = int(input("Enter number to search : "))

if target in nums:
    print(f"{target} exists in {nums}")
else:
    print(f"{target} does NOT exist in {nums}")
```

```python
# Program 3 : Write a program to convert a list into a tuple
nums_list = [10, 20, 30, 40, 50]
nums_tuple = tuple(nums_list)

print(f"nums as a list : {nums_list}")
print(f"nums as a tuple : {nums_tuple}")
```

```python
# Progam 5 : Write a program to0 replace the last of tuples in a list to 100
nums = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
new_list = []

for tup in nums:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)

print(f"Original List : {nums}")
print(f"Updated list : {new_list}")
```

```python
# Progam 6 : Write a program to find the maximum and minimum elment of a tuple
nums = (25, 10, 45, 5, 60, 30)
_max = max(nums)
_min = min(nums)
print(f"Maximum of {nums} is {_max}")
print(f"Minimum of {nums} is {_min}")
```

```python
# Python program to swap the first and last element of a tuple
nums = (10, 20, 30, 40, 50)
new_tuple = (nums[-1],) + nums[1:-1] + (nums[0],)
print(f"Original tuple : {nums}")
print(f"Tuple after swaping : {new_tuple}")
```
