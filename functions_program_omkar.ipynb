```python
def evenOdd(x: int) -> str:
    if x%2 == 0:
        return "Even"
    else:
        return "Odd"

print(evenOdd(16))
print(evenOdd(7))
```

```python
def myFun(x, y=50):
    print(f"x : {x}")
    print(f"y : {y}")

myFun(10)
```

```python
def my_function(country = "Norway"):
    print(f"I am from {country}")

my_function("Sweden")
my_function("India")
my_function()
my_function("Brazil")
```

```python
def student(fname, lname):
    print(fname, lname)

student(fname="Good", lname="Practice")
student(lname="Practice", fname="Good")
```

```python
def nameAge(name, age):
    print(f"Hi, I am, {name}")
    print(f"My age is {age}")

print("Case-1 :")
nameAge("Suraj", 27)

print("")

print("Case-2 :")
nameAge(27, "Suraj")
```
