```python
import pandas as pd
a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)
```

```python
a = [1, 7, 2]
myvar = pd.Series(a, index=['x', 'y', 'z'])
print(myvar)
```

```python
s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'], name="Example series")
print(s1)
```

```python
s2 = pd.Series({'x':1, 'y':2, 'z':3})
print(s2)
```

```python
data = {
    "calories": [420, 380, 390],
    "duration": [50,40,45]
}
myvar = pd.DataFrame(data)
print(myvar)
```

```python
data = {
    'A': [1,2,3,None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1,2,None, None, 5]
}
df = pd.DataFrame(data)
print(f"Original data : \n {df}")
print()

df_cleaned = df.dropna()
print(f"Cleaned data :\n {df_cleaned}")
```

```python
data = {
    'A': [1,2,3,None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1,2,None, None, 5]
}
df = pd.DataFrame(data)
print(f"Original data : \n {df}")
print()

df_filled = df.fillna(0, inplace=False)
print(f"Filled data :\n {df_filled}")
```

```python
data = {
    'A': [1,2,3,None, 5],
    'B': [None, 2, 3, 4, 5],
    'C': [1,2,None, None, 5]
}
df = pd.DataFrame(data)
print(f"Original data : \n {df}")
print()

df_mean = df.fillna(df.mean(), inplace=False)
print(f"Mean values data :\n {df_mean}")
```
