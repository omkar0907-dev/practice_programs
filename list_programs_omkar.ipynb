```python
# counting occurence
nums = [10, 20, 10, 30, 10, 40, 50]
freq = {}
for i in range(len(nums)):
    if freq.get(nums[i], -1) == -1:
        freq[nums[i]] = 1
    else:
        freq[nums[i]] += 1

element = int(input("Enter number to search :"))
if freq.get(element, -1) == -1:
    print(f"{element} NOT found in list")
else:
    print(f"{element} occurs {freq[element]} number of times in list")
```

```python
# Write a program to remove the first occurrence of a specified element from a list
nums = []
n = int(input("Enter number of elements in the list :"))
i = 0
while i < n:
    num = int(input(f"Enter element at index {i} :"))
    nums.append(num)
    i += 1

print("-" * 50)
for i in range(n):
    print(nums[i], end = ", ")
print()
print("-" * 50)

target = int(input("Enter element to remove :"))

pos = -1
for i in range(n):
    if nums[i] == target:
        pos = i
        break

if pos != -1:
    print(f"First occurence of {target} in the list is at index {pos}")
    for i in range(pos, n - 1):
        nums[i] = nums[i + 1]
    nums.pop()
    print("-" * 50)
    print(f"List after removing {target} from index {pos}")
    for i in range(n - 1):
        print(nums[i], end = ", ")
    print()
    print("-" * 50)

else:
    print(f"{target} does NOT occur in the list")
    
```

```python
'''
Accept 20 values from user and save it in list. Perform following operations on it
    a) count similar elements of list and print their index value   
    b) count even and odd values of list 
    c) count positive and negative values of list.
'''
nums = []
n = 20
for i in range(n):
    nums.append(int(input(f"Enter element at index {i}")))

print("\nCount similar elements of list and print their index value")
for i in range(n):
    cnt = nums.count(nums[i])
    if cnt > 1:
        print(nums[i], "->", end=" ")
        for j in range(n):
            if nums[i] == nums[j]:
                print(j, end=" ")
        i += cnt - 1
        print()

print("\nCount even and odd values")
even, odd = 0, 0
for num in nums:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("\nEven values :", even)
print("Odd values :", odd)
```

```python
# Acronym
s = input("Enter string :")
acr = s[0].upper()
for i in range(len(s)):
    if s[i] == " ":
        acr += s[i + 1].upper()
print(f"Acronym for {s} is {acr}")    
```
