```python
marks = int(input("Enter maths marks : "))
grade = ""
if marks >= 90:
    grade = "O"
elif marks >= 80:
    grade = "A+"
elif marks >= 70:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 50:
    marks = "C"
elif marks >= 40:
    grade = "P"
else:
    grade = "F"

print(f"Grade assigned : {grade}")
```

```python
marks = int(input("Enter maths marks : "))
breakpoints = [90, 80, 70, 60, 50, 40, 0]
grades = ["O", "A+", "A", "B", "C", "P", "F"]
for i in range(len(breakpoints)):
    if marks > breakpoints[i]:
        print(f"Grade assigned : {grades[i]}")
        break
```

```python
n = int(input("Enter the number : "))
temp = n
s = 0
while n > 0:
    digit = n % 10
    s += digit
    n //= 10
n = temp
print(f"sum of digits of {n} is {s}")
```

```python
num = int(input("Enter a number :"))
fact = 1
for i in range(1, num + 1):
    fact *= i
print(f"factorial of {num} is {fact}")
```

```python
num = int(input("Enter a number :"))
is_prime = True
for f in range(2, int(num ** 0.5) + 1):
    if num % f == 0:
        is_prime = False
        break
if is_prime:
    print(f"The number {num} is prime")
else:
    print(f"The number {num} is composite")
```

```python
num = int(input("Enter the number :"))
temp = num
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
num = temp
print(f"The reverse of {num} is {rev}")
```

```python
year = int(input("Enter year : "))
if year % 400 == 0:
    print(f"{year} is a leap year")
elif year % 100 == 0:
    print(f"{year} is NOT a leap year")
elif year % 4 == 0:
    print(f"{year} is a leap year")
else:
    print(f"{year} is NOT a leap year")
```
